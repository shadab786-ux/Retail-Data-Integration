# Data cleaning:

```python
#remove duplicate:
df=df.drop_duplicates()
print(df)
```

```python
print(df.duplicated().sum())
```

```python
#for numeric column:
df["Customer_Age"]=df["Customer_Age"].fillna(df["Customer_Age"].median())
print(df["Customer_Age"])
```

```python
#for categorical columns:
df["Customer_Gender"]=df["Customer_Gender"].fillna("unknown")
print(df["Customer_Gender"])
```

# Convert Date Column:

```python
#Convert Invoice_Date into proper datetime format:
df["Invoice_Date"] = pd.to_datetime(df["Invoice_Date"])
print(df["Invoice_Date"])
```

```python
#creating some useful column:
df["Year"] = df["Invoice_Date"].dt.year
df["Month"] = df["Invoice_Date"].dt.month
df["Month_Name"] = df["Invoice_Date"].dt.month_name()
df["Day"] = df["Invoice_Date"].dt.day
df["Day_Name"] = df["Invoice_Date"].dt.day_name()
print("*****Year******")
print(df["Year"])
print("*****Month******")
print(df["Month"])
print("*****Month Name******")
print(df["Month_Name"])
print("*****Day******")
print(df["Day"])
print("*****Day Name******")
print(df["Day_Name"])
```

```python
print(df.columns)
```

# Validate Revenue and Profit:

```python
#Revenue:
df["Calculated_Revenue"] = df["Units"] * df["Selling_Price"]
print(df["Calculated_Revenue"])
```

```python
#Cost:
df["Calculated_Cost"] = df["Units"] * df["Cost_Price"]
print(df["Calculated_Cost"])
```

```python
#Margin
df["Calculated_Margin"] = (
    df["Calculated_Revenue"] - df["Calculated_Cost"]
)
print(df["Calculated_Margin"])
```

```python
#coparing them:
df[["Revenue", "Calculated_Revenue"]].head()
```

# Creating Python Analysis:

# City Analysis:

```python
city_sales = df.groupby("City")["Revenue"].sum().sort_values(
    ascending=False
)
print(city_sales)
```

```python
#top five city revanue:
print(city_sales.head())
```

# Category Analysis:

```python
category_sales = df.groupby("Category")["Revenue"].sum().sort_values(
    ascending=False
)
print(category_sales)
```

```python
#checking profit category wise:
category_profit = df.groupby("Category")["Margin"].sum().sort_values(
    ascending=False
)
print(category_profit)
```

# Brand Analysis:

```python
brand_sales = df.groupby("Brand")["Revenue"].sum().sort_values(
    ascending=False
)
print(brand_sales)
```

```python
#top 10 brand:
print(brand_sales.head(10))
```

# Online vs Offline:

```python
channel_sales = df.groupby("Channel")["Revenue"].sum()
print(channel_sales)
```

# Payment Mode Analysis:

```python
payment_sales = df.groupby("Payment_Mode")["Revenue"].sum()
print(payment_sales)
```

# Monthly Sales Trend:

```python
monthly_sales = df.groupby("Month")["Revenue"].sum()
print(monthly_sales)
```

```python
#For a cleaner result:
monthly_sales = df.groupby(
    df["Invoice_Date"].dt.to_period("M")
)["Revenue"].sum()
print(monthly_sales)
```

# Customer Analysis:

```python
gender_sales = df.groupby("Customer_Gender")["Revenue"].sum()
print(gender_sales)
```

```python
#Loyalty customers:
loyalty_sales = df.groupby("Loyalty_Flag")["Revenue"].sum()
print(loyalty_sales )
```
